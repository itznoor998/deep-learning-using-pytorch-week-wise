# The Training Loop (Core of All DL)

**The one loop that rules everything** <br> 
```python
for each batch:
    zero gradients
    forward pass
    compute loss
    backward pass
    update weights
```

````python
dtype = torch.float 
device = torch.cuda
````

## Minimal training (No dataset yet)

````python
torch.manual_seed(22)

x = torch.tensor([1,2,3,4], dtype=dtype) 
y_true = torch.tensor([3,5,7,9], dtype=dtype) 

class LinearModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.w = nn.Parameter(torch.randn(1)) 
        self.b = nn.Parameter(torch.randn(1)) 

    def forward(self,x): 
        y = self.w * x + self.b 
        return y 
    

model = LinearModel() 
loss_fn = nn.MSELoss() 
optimizer = torch.optim.SGD(model.parameters(), lr = 0.1)

print('w',model.w,', b',model.b)
````

## One trainign step (Manual)

````python
y_pred = model(x) 
loss = loss_fn(y_true,y_pred) 

optimizer.zero_grad() 
loss.backward() 
optimizer.step()

print("Loss:",loss.item()) 
print("w:",model.w.item(), ", b:",model.b.item())
````

- What happened to `w` and `b`?

- Why was `zero_grad()` before `backward()`?

## Turning it into a loop

````python
epochs = 20 

for epoch in range(epochs): 
    # Forward pass
    y_pred = model(x) 
    # Compute loss
    loss = loss_fn(y_true,y_pred) 

    # Zero gradients
    optimizer.zero_grad()
    # Backward pass
    loss.backward() 
    # Weights update
    optimizer.step()

    print(f"Epoch: {epoch+1}, Loss: {loss.item():.4f}, w: {model.w.item():.4f}, b: {model.b.item():.4f}")
````

- What happens to loss?

- Where is “learning” happening?

## Understand the gradient flow

````python
torch.manual_seed(33)
epochs = 10 
model = LinearModel() 
loss_fn = nn.MSELoss() 
optimizer = torch.optim.SGD(model.parameters(), lr = 0.1)

print('w',model.w,', b',model.b,'\n\n')




for epoch in range(epochs): 
    # Forward pass
    y_pred = model(x) 
    # Compute loss
    loss = loss_fn(y_true,y_pred) 

    # Zero gradients
    optimizer.zero_grad()
    # Backward pass
    loss.backward() 
    # Weights update
    optimizer.step()

    print(f"lr * grad w: {0.1*model.w.grad.item():.4f}, lr * grad b: {0.1*model.b.grad.item():.4f} \n")
    print(f"Epoch: {epoch+1}, Loss: {loss.item():.4f}, w: {model.w.item():.4f}, b: {model.b.item():.4f}")
````

- What do these numbers represent?

- Why do they change each epoch?

# Interaction

### 1️⃣ Why `zero_grad()`


- `zero_grad()` does not free the graph.
  
- It clears the `.grad` buffers of parameters.


Graph is freed by backward() automatically.

Gradients live in parameter buffers.

So:
```python
loss.backward()   # builds grads
optimizer.zero_grad()   # clears old grads
```

***

### 2️⃣ Why gradients accumulate (the real reason)



Inside PyTorch:
```bash
param.grad = param.grad + new_gradient
```
NOT
```bash
param.grad = new_gradient
```
So if you skip `zero_grad()`:

| Epoch | Gradient stored |
| ----- | --------------- |
| 1     | g₁              |
| 2     | g₁ + g₂         |
| 3     | g₁ + g₂ + g₃    |
| …     | keeps growing   |


That’s gradient accumulation.

### 3️⃣ Why loss explodes when we skip `zero_grad()`


Let’s see what optimizer does:

$𝑤 = 𝑤 − \eta⋅grad$


If grad = g₁ + g₂ + g₃ + …

Then:

- Update becomes huge

- Model overshoots

- Loss increases instead of decreases

This is not “true exploding gradients”.
- It is wrong gradient accumulation.

### 4️⃣ Exploding gradients vs accumulated gradients


| Exploding gradients        | Accumulated gradients               |
| -------------------------- | ----------------------------------- |
| Comes from deep nets, RNNs | Comes from forgetting `zero_grad()` |
| Mathematical instability   | Programming mistake                 |
| Solved by clipping         | Solved by zeroing                   |


`zero_grad()` = accumulation bug, not mathematical explosion.

### 5️⃣ Why `.grad` changes every epoch


As `w` and `b` move closer to solution:

- Loss curve flattens

- Gradients shrink

- Model converges

This is optimization in action.

## What is Transfer Learning?

A model like **ResNet** is trained on **ImageNet**:

- 1M+ images

- 1000 classes

- Animals, vehicles, objects, textures, etc.

During this training it learns:

- Edges

- Curves

- Patterns

- Object parts

We reuse this knowledge.

## Two Ways to Use Transfer Learning

---

### Method 1 — Feature Extraction


- Freeze all convolution layers

- Only train the final classification layer

Use when:

- Dataset is small

- Problem is similar

Example:

- Pretrained on ImageNet

- Your task: classify fruits

Edges, shapes → already learned.

---

### Method 2 — Fine-Tuning

- Freeze early layers

- Train last few layers

Use when:

- Dataset is medium-sized

- Problem is a bit different

# Interacion

### Why training from scratch is hard


- Needs lots of data ✔

- Time-consuming ✔

- Computationally expensive ✔

- Deep CNNs have millions of parameters. Without enough data, they overfit badly ✔

---

###  Why freeze early layers


Retraining early layers will make them forget previous patterns/knowledge.

This phenomenon called:

- **Catastrophic forgetting**

Also:

- Early layers already learned universal visual features

- Retraining them may destroy useful general features

So freezing:

- Saves compute

- Preserves learned knowledge

- Prevents overfitting on small datasets

---

### Early vs deep layer reuse 

Vehicle → animal example
- Let's assume we have a pretrained model to classify between vehicles. 
- The initial layers of the CNN learns edges, lines, strokes, color transitions. 
- And the later layers sees the object or a part of a object. 
- Now we want to reuse the model to classify dog, cat and others animals. 
- So the learnings of the initial layers will be useful but the learnings of the later or deeper layers may not be useful. 
- Like a part of a vehicle migh not similar to a part of a animal but we need edges, curves, small strokes to form a animal.

```java
Early layers → general (edges, strokes) ✔

Deep layers → task-specific (car parts vs animal parts) ✔
```

### Confusion 1 

“How do we decide from which layer to start training?”

There is no single fixed rule. But there is a very reliable practical strategy.

Let's think in terms of dataset size + similarity.

#### Case A — Small dataset (most common)


Example:

- 1000–5000 images

Do this:

- Freeze ALL convolution layers

- Train only final classifier

Why?

- Too little data to safely retrain deep layers

- Avoid overfitting

---

#### Case B — Medium dataset



Example:

- 10k–50k images

Do this:

- Freeze early layers

- Train last few conv layers + classifier

Why?

- Early layers still useful

- Later layers need adaptation

---

#### Case C — Very large dataset



Example:

- 100k+ images

Do this:

- Fine-tune entire network

Why?

- Enough data to relearn everything

---

#### In-short
| Dataset size | What to train   |
| ------------ | --------------- |
| Small        | Only last layer |
| Medium       | Last few layers |
| Large        | Full network    |

### Confusion 2



“Can I reuse any pretrained CNN for any task?”

Short answer:

- Mostly **yes** for vision tasks, but not always equally good.


---

#### When reuse works well



If original training was on:

- ImageNet (animals, objects, textures)

We can reuse for:

- Medical images

- Fruits

- Vehicles

- Faces

- Plants

- Defects

Because early visual features are universal.

---

#### When reuse is weaker

If new task is VERY different:

Examples:

- Satellite images

- X-ray scans

- Microscopy cells

Still works, but:

- Need more fine-tuning

- Sometimes train deeper layers

---

#### When reuse is NOT ideal



If data type is completely different:

Example:

- Spectrograms

- Non-image matrices

Then CNN may still work, but pretrained ImageNet weights help less.

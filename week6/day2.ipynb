## Why Constant Learning Rate Is Not Ideal

If learning rate is constant:
```bash
Early training → too small

Late training → too large
```

Why?

Early:

- We are far from optimum

- Large steps help


Late:

- We are near optimum

- Large steps overshoot


So ideal behavior:

- Large LR initially
- Small LR near convergence

## Step Decay



Classic approach:

- Reduce LR after certain epochs.

Example:
```python
Epoch 0–30 → 0.1
Epoch 30–60 → 0.01
Epoch 60–90 → 0.001
```
Pros:

- Simple

- Works well


Cons:

- Manual tuning

- Sudden jumps

## Exponential Decay



Smooth reduction:

$\eta_t = \eta_0 \cdot e^{-kt}$

Smooth but:

- Hard to tune k

## Cosine Annealing (Modern Favorite)

Learning rate follows cosine curve:
```bash
High → gradually decreases → near zero
```
Advantages:

- Smooth

- No sudden jumps

- Works very well in practice


Widely used in:

- Vision models

- Transformers

## Warmup (Very Important) 

In very deep models:

Starting with large LR can:

- Destabilize training

- Cause exploding gradients


So we:

1. Start with small LR


2. Gradually increase (warmup)


3. Then decay



This is called:

> Warmup + decay schedule



Used in:

- BERT

- GPT

- Large CNNs

## Why Scheduling Helps Generalization

Large LR early:

- Escapes sharp minima

- Explores loss landscape


Small LR late:

- Fine-tunes precisely


This often leads to:

> Flatter minima → better generalization

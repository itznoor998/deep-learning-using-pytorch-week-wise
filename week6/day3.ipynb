# Sharp vs Flat Minima (Why Some Models Generalize Better)

**Theme:**

*“Not all minima are equal.”*

Two models can:

- Have same training loss

- But very different test performance


Why?

- The answer often lies in the shape of the minimum.

## Visual Intuition

Imagine 2 valleys in loss landscape.

Sharp Minimum
```bash 
id='sharp-minima'

\__/
       \/
```
Very narrow bottom.
```bash
Small change in parameters → big increase in loss.
```

---

Flat Minimum
```bash
id='flatter-minima'

\_____/
      \____/
```
Wide bottom.
```bash
Small parameter change → little change in loss.
```

## Why This Matters



Neural networks:

- Have millions of parameters

- Training data is noisy

- Initialization is random


If model lands in sharp minimum:

- Small perturbations hurt performance

- Poor generalization


If model lands in flat minimum:

- Robust to perturbations

- Better generalization

## Why Learning Rate Matters Here

Large learning rate:

- Cannot settle into narrow sharp valleys

- Jumps out of them

- Tends to land in flatter regions


Small learning rate:

- Can descend into sharp minima

- Very precise


This is why:

> Large LR early → better exploration

> Small LR late → refinement


## Why SGD Often Generalizes Better Than Adam



SGD:

- Noisy updates

- Bounces around

- Hard to settle in sharp minima


Adam:

- Smooth updates

- Quickly converges

- Can settle in sharper minima


This explains why:

- SGD sometimes generalizes better.

## Real-World Insight



Flat minima are correlated with:

- Better robustness

- Better generalization

- Better performance on new data


This is an active research topic.

## Reminder: Basic Gradient Descent

$w = w - \eta \nabla L(w)$

Where:

$\eta =$ learning rate


This is simple.

But it has problems.

##  Problems of Plain SGD

### 1️⃣ Oscillation in Narrow Valleys



Imagine loss surface like:
```bash
steep
        |
        |
        V
       \____/
```
1. One direction steep
2. Other direction flat

SGD:

- Bounces left-right

- Slow convergence



---


### 2️⃣ Sensitive to Learning Rate

- Too large → divergence
- Too small → very slow


---

### 3️⃣ No Memory

Each update only depends on:

- Current gradient


It ignores:

- Past direction

- Smoothing

## Momentum (First Major Improvement)


Momentum adds memory:

$v_t = \beta v_{t-1} + \nabla L$ <br>
$w = w - \eta v_t$

So:

- Gradients accumulate

- Noise smooths out

- Convergence faster



Intuition:

  Think of a ball rolling downhill:

- Without momentum → jittery

- With momentum → smooth rolling

## Why Adam Became Popular

Adam combines:

- Momentum

- Adaptive learning rates


It adjusts learning rate per parameter.

So:

- Works well out of the box

- Less tuning needed

- Good for deep networks

## Output Size Formula (VERY IMPORTANT SKILL)

You don’t need to memorize deeply, just understand logic.

For conv layer:

**Output = $\frac{W-K+2P}{S} + 1$**

Where:

- W = input size

- K = kernel size

- P = padding

- S = stride

# Interaction

### Stride 



Professional phrasing:
```bash
Stride = 1 → filter moves 1 pixel at a time

Stride = 2 → filter jumps 2 pixels at a time

Larger stride → fewer positions → smaller output
```

---

### Padding 


Padding helps preserve spatial resolution so deeper networks don’t shrink too fast.

If we never used padding:
```bash
28 → 26 → 24 → 22 → ...
```
After many layers, the feature map would disappear.

### Stride vs Pooling

```bash
Stride reduces size by skipping locations

Pooling reduces size by summarizing a region
```
Pooling keeps:

- strongest features

Stride just:

- samples less frequently

So pooling is more meaningful for vision.

# Interaction

### Why flattening is bad 

- Flattening destroys the spatial relationships between pixels, which are critical for recognizing shapes, edges, and patterns in images.

- A CNN preserves local neighborhoods.

---

### Why filters are reused

The same filter is reused across the image because a pattern (like an edge or curve) can appear anywhere, and we want to detect it regardless of location.

This gives:

- **Translation invariance**

- Fewer parameters

- Better generalization

---

###  Weight sharing

Weight sharing means the same filter (weights) is applied to different spatial locations in the same layer — not across layers.

So:

- One filter

- Slides across image

- Detects same pattern everywhere

---

### CNN vs MLP

- Key difference:

CNNs exploit spatial structure, MLPs ignore it.

---

### “How does CNN learn low-level → high-level features?”

> This is a fundamental deep learning insight.


#### Layer 1 — Low-level features



First convolution layer sees raw pixels.

It learns filters that detect:

- Edges

- Corners

- Light vs dark

- Simple textures

Why?

Because these patterns:

- Appear everywhere

- Are useful building blocks

So filters become edge detectors.

---

#### Layer 2 — Mid-level features


- Layer 2 does not see pixels.
- It sees edges from layer 1.

So it combines edges into:

- Curves

- Corners

- Simple shapes

- Strokes in digits

---

#### Layer 3+ — High-level features 


Now CNN sees shapes of shapes.

For MNIST, these become:

- Loops

- Lines

- Crossings

- Digit-specific patterns

Eventually:

- “This looks like a 9”

- “This looks like a 4”

---

#### Why this happens naturally



Because:

- Lower layers are forced to explain raw data

- Higher layers explain patterns of patterns

Mathematically:

- Each layer learns features that make the next layer’s job easier.

- This is called hierarchical feature learning.

> One-sentence interview answer

CNNs learn low-level features in early layers and high-level features in later layers because each layer composes simpler patterns into more complex ones.

# Imports

````python
import torch 
import torch.nn as nn

torch.__version__
````

````python
from torchvision import datasets, transforms 
from torch.utils.data import DataLoader, random_split   
````

# Optimizer, Learning rates and Why training fails

## Try with a high learning rate (Break the model)

````python
transform = transforms.ToTensor() 

train_data_full = datasets.MNIST(root='data', train=True, download=True, transform=transform)  

train_size = int(0.8 * len(train_data_full))   
val_size = len(train_data_full) - train_size 

train_data, val_data = random_split(train_data_full, [train_size, val_size])  

train_loader = DataLoader(train_data, batch_size=64, shuffle=True) 
val_loader = DataLoader(val_data, batch_size=64, shuffle=False) 
````

````python
# Neural Network
class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Flatten(),
            nn.Linear(28*28, 128),
            nn.ReLU(),
            nn.Linear(128,10),
        )
    
    def forward(self, x):
        return self.net(x) 

model = MLP()
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1.0)  
````

📌 Observe:

- Train loss

- Validation loss

Write what happens.

````python
for epoch in range(10):
    # TRAIN 
    model.train()
    train_loss = 0.0 
    for images, labels in train_loader:
        # forward pass 
        outputs = model(images) 
        # compute loss 
        loss = loss_fn(outputs, labels) 

        # zero gradient 
        optimizer.zero_grad() 
        # backward pass
        loss.backward() 
        # update parameters 
        optimizer.step()

        train_loss += loss.item() 

    # VALIDATION 
    model.eval() 
    val_loss = 0.0 
    with torch.no_grad():
        for images, labels in val_loader:
            outputs = model(images) 
            loss = loss_fn(outputs, labels) 

            val_loss += loss.item() 

    print(f"Epoch: {epoch+1} | Train Loss: {train_loss/len(train_loader):.4f} | Validation Loss: {val_loss/len(val_loader):.4f}")
````

## Try with a very low learning rate

````python
# Model, loss function, optimizer
model2 = MLP()
loss_fn2 = nn.CrossEntropyLoss()
optimizer2 = torch.optim.Adam(model2.parameters(), lr=0.00001)  
````

📌 Observe:

- How fast does loss decrease?

- Compare with lr = 0.001

````python
for epoch in range(10):
    # TRAIN 
    model2.train()
    train_loss = 0.0 
    for images, labels in train_loader:
        # forward pass 
        outputs = model2(images) 
        # compute loss 
        loss = loss_fn2(outputs, labels) 

        # zero gradient 
        optimizer2.zero_grad() 
        # backward pass
        loss.backward() 
        # update parameters 
        optimizer2.step()

        train_loss += loss.item() 

    # VALIDATION 
    model2.eval() 
    val_loss = 0.0 
    with torch.no_grad():
        for images, labels in val_loader:
            outputs = model2(images) 
            loss = loss_fn2(outputs, labels) 

            val_loss += loss.item() 

    print(f"Epoch: {epoch+1} | Train Loss: {train_loss/len(train_loader):.4f} | Validation Loss: {val_loss/len(val_loader):.4f}")
````

## Compare Optimizers

````python
model3 = MLP()

optimizer3 = torch.optim.SGD(model3.parameters(), lr=0.1) # now with SGD instead of Adam
````

````python
for epoch in range(10):
    # TRAIN 
    model3.train()
    train_loss = 0.0 
    for images, labels in train_loader:
        # forward pass 
        outputs = model3(images) 
        # compute loss 
        loss = loss_fn2(outputs, labels) 

        # zero gradient 
        optimizer3.zero_grad() 
        # backward pass
        loss.backward() 
        # update parameters 
        optimizer3.step()

        train_loss += loss.item() 

    # VALIDATION 
    model3.eval() 
    val_loss = 0.0 
    with torch.no_grad():
        for images, labels in val_loader:
            outputs = model3(images) 
            loss = loss_fn2(outputs, labels) 

            val_loss += loss.item() 

    print(f"Epoch: {epoch+1} | Train Loss: {train_loss/len(train_loader):.4f} | Validation Loss: {val_loss/len(val_loader):.4f}")
````

# Interaction

### our results are spot-on ✅


`lr = 1.0`

- Loss stuck around `~2.3`

- ➡ That’s ≈ log(10) → random guessing

- Meaning: optimizer is jumping around, not learning.

`lr = 0.00001`

- Loss decreases, but painfully slow
- ➡ Model can learn, but training is inefficient.

Adam vs SGD

- Despite higher LR, Adam worked decently
- ➡ Because Adam internally adapts the learning rate per parameter, while SGD relies purely on our LR choice.

---

### The professional default (industry rule)



For most problems, start with:
```python
optimizer = Adam(model.parameters(), lr=1e-3)
```
Why?

- Works on most datasets

- Robust to bad initialization

- Requires minimal tuning

This is why Adam is used in 90% of research code.

---

### Learning rate finding (practical method) 

#### Method 1 — Try 3 values (most common)

Always try:

| Learning rate | Meaning       |
| ------------- | ------------- |
| `1e-1`        | Too high test |
| `1e-3`        | Safe default  |
| `1e-5`        | Too low test  |


Pick the one that:

- Decreases loss fastest

- Is stable

We already did this instinctively. That’s good.

---

#### Method 2 — LR Range Test (advanced, optional) 

Gradually increase LR during training:

- Track loss

- Choose LR just before loss explodes

Used in:

- fast.ai

- Large-scale training

We’ll learn this later.

### Choosing the optimizer (clear rules)


Use Adam when:

- Dataset is small/medium

- Features are noisy

- We want fast convergence

- We are prototyping

Use SGD (+ momentum) when:

- Dataset is very large

- We care about final generalization

- We are training CNNs from scratch

> Interview answer:

Adam converges faster, SGD often generalizes better.

---

### Why SGD worked with lr = 0.1

SGD update is simple:

$𝑤 = 𝑤 − \eta\cdot∇𝑤$

MNIST is:

- Simple

- Well-conditioned

- Clean

So SGD behaves well.

**On messy data, Adam usually wins early.**

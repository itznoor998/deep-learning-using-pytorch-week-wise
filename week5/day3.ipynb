# Residual Connections (The Idea That Made Very Deep Networks Possible)

**Theme:**

 *“If learning becomes hard, let information skip layers.”*

Before ResNet:
```bash
10–20 layers → OK

50+ layers → training failed
```
Loss stopped improving

Gradients vanished


Researchers expected:

> Deeper network = better performance



But they observed:

> Deeper network = WORSE performance



This was shocking.

## The Core Problem



Imagine a deep stack:
```bash
Layer1 → Layer2 → Layer3 → ... → Layer50
```
Backprop must travel through:

- 50 transformations

- 50 multiplications

- 50 nonlinearities


Gradients become:

- Very small

- Very distorted


So early layers stop learning.

## The Brilliant Ideas 

ResNet introduced a simple trick:

Instead of learning:

$y = F(x)$

Let the layer learn:

$y = F(x) + x$

This means:

- The input skips the layer

- Gets added to the output


This is called:

> Skip Connection or Residual Connection

## Intuition (Very Important)

Suppose a layer is not useful.

Without skip connection:

- It still modifies the signal

- May damage information


With skip connection:

Model can learn:

$F(x) = 0$<br> Then,
$y = x$

Meaning:

> The layer becomes an identity layer.


So deeper networks stop hurting performance.

## Why It Fixes Vanishing Gradients

Backprop now has two paths:

1. Through the layer


2. Direct skip path



So gradient can flow:
```bash
directly → to earlier layers
```
This keeps gradients alive.

This is why:

- 50 layers

- 101 layers

- 152 layers

became trainable.

## Simple Mental Model

Without skip:
```bash
Information → squeezed → squeezed → squeezed
```
With skip:
```bash
Information → processed
         ↘ skip ↗
```
So original information is preserved.

# Answers

### Q1 — Why deep networks failed before ResNet



- Many transformations

- Gradients becoming very small

- Early layers stopping learning

That is exactly the vanishing gradient problem in deep stacks.

---

### Q2 — Meaning of 
$y=F(x)+x$



Interpretation:

Instead of learning a completely new transformation, the layer only needs to learn:

- “What extra change should I add to x?”

So the layer learns a residual (difference), not the whole mapping.

---

### Q3 — Why skip helps gradient flow



- One path may shrink gradients

Skip path gives a direct route

- So gradients can still reach earlier layers.

---


### Q4 — If layer is useless



Network ignores the layer


Mathematically:
- If the best solution is:

$F(x)=0$

- Then:

$y=x$

So the network behaves like:

- That block doesn’t exist.

This is why deeper networks stopped getting worse.

---

### Q5 — Which is easier to learn?



Learning:

$y=F(x)+x$

is easier than learning:

$y=H(x)$

Why?

- Because if the best solution is identity:

  - Just learn, 
  $F(x)=0$, means "do nothing"

- That’s easier than forcing a stack of layers to learn identity from scratch.

So residual form makes optimization easier.

# Notes

### How ResNet Really Uses Skip Connections



Skip connections are NOT added after every single layer.

They are added after blocks.

A typical ResNet block:
```ruby
Conv → BN → ReLU →
Conv → BN
        + skip connection
        ↓
       ReLU
```
So the skip connection spans:

`2 layers`
or

`3 layers (in deeper versions)`

So structure is more like:
```ruby
x → [Layer1 → Layer2] → +x
```
Not:
```ruby
x → Layer1 → +x → Layer2 → +x
```

### Why Skip Every Few Layers?



Because:

- One layer is too small to learn meaningful residuals

- A small group of layers can learn a useful transformation

- Skip path ensures original signal still flows

So residual block = a mini network + shortcut.

---

### Can We Skip Every 3rd Layer?



***Yes.***

And many architectures do variations like:

- DenseNet

- Highway Networks

- U-Net

They all play with different skip structures.

But ResNet found a very stable pattern:
```js
Skip every 2–3 layers → works best.
```
--

### Intuition



Think of it like this:

Instead of saying:

- “Fix the signal after every step”

We say:

- “Fix the signal after a small processing unit.”

So the block learns:

- **Refinement of x**
***

### Big Insight You Should Remember



Residual connections are not about skipping layers randomly.

They are about:

- Letting information and gradients move easily through depth.

That’s why we can train:

- 50 layers

- 101 layers

- 152 layers

Without collapse.

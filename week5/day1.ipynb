# Vanishing & Exploding Gradients (THE Root Problem)

This is one of the most important concepts in deep learning.

**Theme:**

*"Deep networks don’t fail because they’re dumb.<br>They fail because the learning signal can’t survive."*

## What is a Gradient Again?

During backprop:

Each layer updates weights using:
    
$$
w = w -  \alpha \frac {\partial L}{\partial w}
$$
- $\alpha$ is learning rate

So training depends on:

- How large the gradients are.

## The Two Big Problems

### 1) Vanishing Gradient



Gradients become:
```bash
very very small → almost 0
```

Then:

- Early layers stop learning

- Network learns very slowly

- Deep networks fail

### 2) Exploding Gradient



Gradients become:
```bash
very very large
```

Then:

- Weights jump wildly

- Loss becomes unstable

- Training crashes

## Why This Happens (Intuition)



Imagine 10 layers.

Backprop multiplies gradients layer by layer:
```bash
g × 0.5 × 0.5 × 0.5 × ...
```

After many multiplications:
```bash
≈ 0
```

→ Vanishing

OR
```bash
g × 2 × 2 × 2 × ...
```

After many layers:
```bash
very large
```

→ Exploding

## Why Shallow Networks Worked Earlier



Our MNIST CNN:

- Only few layers

So:

- Gradient didn't vanish

- Training was stable

But in deep networks:

- 50, 100, 150 layers

This becomes a serious problem.

# Answers

### Q1 — If early layers stop learning


If early layers stop learning:

- They remain stuck with poor filters

- Later layers get weak/low-quality features

- Final predictions become limited

So the network can’t learn strong representations.

---

### Q2 — Why deeper networks suffer more

Multiplication of small numbers becomes zero

Core mathematical intuition:
- Repeated multiplication of numbers < 1 shrinks gradients.

---

### Q3 — Which is worse?

In practice:

- Vanishing gradient is more common

- Exploding gradient is more dramatic but easier to detect

Why?

- Exploding:

  - Loss becomes NaN

  - Training crashes

- Vanishing:

  - Training continues

  - But model silently stops improving

So vanishing is more dangerous because it’s subtle.

---

### Q4 — Detecting vanishing gradient

Signs:

- Loss plateaus early

- Accuracy stuck

- Early layers barely change

# Confusion

> How to choose learning rate or weights to minimize vanishing/exploding gradients?

The answer is:

- We don’t rely on only LR or weights.

- Modern deep learning solved this problem using 4 major techniques.

The Real Solutions Used Today


### 1️⃣ Weight Initialization (Very Important)


Instead of random numbers like:

- Normal(0, 1)

We use smart initialization like:

- **Xavier** initialization

- **He** initialization

These are designed to:

- Keep gradients stable across layers

- So gradients neither shrink nor explode.

PyTorch already does this internally.

---

### 2️⃣ Activation Functions


Old activation:

- Sigmoid / Tanh

These caused severe vanishing gradients.

Modern choice:

- ReLU

Why ReLU helps:

- Gradient = 1 for positive inputs

- Doesn’t shrink gradients much

This is one reason ReLU became dominant.

---

### 3️⃣ Batch Normalization (Next topic)

BatchNorm:

- Normalizes activations

- Keeps values in stable range

- Makes gradient flow smoother

This is a major breakthrough in deep learning.

*We’ll study this in **Week 5 Day 2**.*

---

### 4️⃣ Residual Connections (ResNet)


This was a game changer.

Instead of:
```js
Layer1 → Layer2 → Layer3
```
ResNet adds:
```css
Input → skip → Output
```
This allows gradients to flow directly.

This is why we can train:

- 50-layer networks

- 100-layer networks

Without vanishing gradients.

---

### What About Learning Rate?


Learning rate mostly affects:

- Speed of training

- Stability

But it is NOT the main cause of vanishing gradients.

However:
```ruby
Too large LR → exploding behavior
Too small LR → slow learning
```
But the real vanishing gradient problem comes from:

- Deep networks

- Poor activations

- Bad initialization

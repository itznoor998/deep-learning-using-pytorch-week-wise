# Activation Functions (Why ReLU Changed Deep Learning)



**Theme:**

 *“Without activation, a deep network is just a linear equation.”*

## Why Do We Even Need Activation Functions?

Imagine a network:
```bash
Linear → Linear → Linear → Linear
```
Mathematically, this becomes:

$y = W_4(W_3(W_2(W_1x)))$

Which simplifies to:

$y = Wx$

So no matter how many layers:

> It’s still just one linear transformation.



This means:

- Cannot learn complex patterns

- Cannot model curves

- Cannot separate non-linear data


So we need:

> Non-linearity



That’s the job of activation functions.

## Old Activations (Sigmoid, Tanh)


### Sigmoid

$\sigma(x) = \frac{1}{1 + e^{-x}}$

Output range:

- 0 to 1

Problems:

- Saturates at extremes

- Gradient becomes very small

- Causes vanishing gradients

### Tanh

Output range:

- -1 to 1

Better than sigmoid because:

- Centered around 0


But still:

- Suffers from vanishing gradient

## ReLU (The Game Changer)


$\text{ReLU}(x) = \max(0, x)$

So:

- If x > 0 → output = x

- If x ≤ 0 → output = 0


Why it changed deep learning:

1. No saturation for positive values


2. Gradient = 1 for positive region


3. Very simple computation


4. Much less vanishing gradient



This allowed:

- Deeper networks

- Faster training

## The "Dying ReLU" Problem

ReLU has one weakness:

If neuron gets stuck in negative region:
```bash
x < 0 → output = 0
gradient = 0
```
Then:

- It may never recover

Neuron becomes inactive forever


This is called:

> Dying ReLU

## Modern Variants



To fix dying ReLU:

- Leaky ReLU

Instead of:

- 0 for negative

Use:

- small slope (e.g., 0.01x)

So neuron never fully dies.

# Answer

### Q1 — Linear stacking

Stacking linear layers:

$W_3(W_2(W_1x)) = Wx$

So without activation:

- Depth adds no expressive power.

---

### Q2 — Why sigmoid/tanh cause vanishing gradients

When input is large positive or negative:

- Derivative ≈ 0

- Gradient becomes tiny

- Backprop multiplies many small numbers

Vanishing happens

---

### Q3 — Why ReLU helps

- No saturation in positive region

- Derivative = 1

- Cheap to compute

Enables deeper networks

---

### Q4 — Dying ReLU

If neuron outputs negative consistently:

- Gradient = 0

- It never updates again

So, the neuron dies forever.

---

### Q5 — Gradient=1 vs 0.001

- Multiplying 1 repeatedly keeps signal alive.

- Multiplying 0.001 repeatedly kills it.

# Confusion

### Confusion 

> There are activation functions better than ReLU in theory.<br>
  Why is ReLU still the most used?

Let’s break it down carefully.



#### 1️⃣ “Better in theory” $≠$ “Better in practice”

Many activations are proposed:

- Leaky ReLU

- ELU

- SELU

- GELU

- Swish

- Mish

In theory they:

- Are smoother

- Avoid dying ReLU

- Have nicer derivatives

But in practice:

- ReLU is simple, stable, and good enough.

And in engineering:

- Simple + stable often wins.



---

2️⃣ ReLU Is Computationally Cheap

ReLU:

$max(0,x)$

That’s it.

- No exponentials.
- No divisions.
- No expensive math.

On GPUs:

- Extremely fast

- Easy to optimize

Speed matters a lot at scale.

---

3️⃣ ReLU Works Well with Modern Techniques

Deep networks today use:

- **He** initialization

- BatchNorm

- Residual connections

These already:

- Reduce vanishing gradients

- Improve stability

So ReLU’s weaknesses are less severe now.

---

#### 4️⃣ Dying ReLU Is Not That Catastrophic in Practice

In large networks:

- Even if some neurons die,

- Many others remain active.

So overall model still works fine.

---

#### 5️⃣ Industry Standardization

Most famous architectures use ReLU:

- VGG

- ResNet

- EfficientNet (mostly variants)

So people stick with:

- What is proven and robust.

---

### When Do People Use Other Activations?



Now here’s the important nuance.

Modern large models often use:

- GELU

  - Used in:

    Transformers

    BERT

    GPT

  - Because:

    Smoother

    Better for attention models

- Swish

  - Used in:

    EfficientNet

  - Because:

    Slight accuracy improvement

So ReLU is dominant in:

- CNN-based vision models

But in NLP:

- GELU dominates

---

### Big Insight

ReLU is not “the best mathematically.”

It is:

- The best trade-off between simplicity, speed, and performance.

And in real-world ML:

- Trade-offs matter more than theoretical elegance.

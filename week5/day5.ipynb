## What Happens If We Initialize Poorly?


Imagine:

### Case A — Weights too small
```python
W ≈ 0.0001
```
Then:

- Outputs become tiny

- Activations shrink

- Gradients shrink

- Vanishing gradient problem



### Case B — Weights too large
```python
W = 10
```
Then:

- Outputs explode

- Activations explode

- Gradients explode

- Training unstable

So initialization scale is CRITICAL.

## The Goal of Good Initialization



We want:

- Forward pass variance stable

- Backward pass gradients stable

- No explosion

- No vanishing

So the variance of outputs should stay roughly constant across layers.

## Xavier Initialization (Glorot)



Designed for:

- Sigmoid

- Tanh

Idea:

Keep variance same across layers.

Formula (intuition only):
$$
Var(W) = \frac{1}{n_{in}}
$$

Where:

#### $n_{in} =$ number of input neurons

This keeps signals stable.

## He Initialization (Very Important)



Designed for:

- ReLU

Why different?

Because ReLU:

- Zeros out half the activations

So effective variance is reduced.

He initialization compensates for that:
$$
Var(W) = \frac{2}{n_{in}}
$$


This is why ReLU + He initialization works so well together.

## Why This Matters Deeply

Without proper initialization:

- Deep networks fail to train

- Gradients vanish or explode

- Learning becomes unstable

With good initialization:

- Stable gradient flow

- Faster convergence

- Deep training becomes feasible

## PyTorch Already Does This

When you define:
```bash
nn.Conv2d(...)
nn.Linear(...)
```
PyTorch automatically applies:

- He initialization for ReLU-based layers

So you’ve been using it without realizing.

# Answers

### Q1 — Why zero initialization fails

The main reason zero initialization fails is actually:

> Symmetry problem



If all weights are initialized to zero:

- Every neuron in a layer produces the same output

- Every neuron receives the same gradient

- They all update identically


So the network behaves like:

- One neuron copied many times



This means:

- No diversity

- No feature learning

- Model collapses


So zero initialization fails mainly because of symmetry, not variance.


---

### Q2 — Very small weights → vanishing

```css
Small weights → small activations → small gradients → repeated multiplication → gradient shrinks exponentially.
```


---

### Q3 — Why ReLU needs different initialization

- ReLU zeros negatives so effective variance reduces


- Since ReLU kills about half the values, variance reduces by roughly half.

- So **He** initialization compensates by doubling variance:
$$
Var(W) = \frac{2}{n_{in}}
$$

---

### Q4 — Why variance preservation matters

If variance grows layer by layer:

- Activations explode


If variance shrinks:

- Activations vanish


So we want:

- Variance roughly constant across layers.



That keeps both forward and backward passes stable.

# Confusion


> Why don’t we just manually set some weights positive and some negative?


Short answer:

- That is exactly what good initialization methods already do.



Let’s go deeper.


---

1️⃣ Modern Initializations Already Use Positive & Negative Values

When PyTorch initializes weights:

- They are sampled from a distribution centered at 0

Typically:

- Normal distribution

- Uniform distribution


With carefully chosen variance


So weights are:

- Some positive

- Some negative

- Balanced around zero


That’s critical.


---

2️⃣ Why Centering Around Zero Matters

If all weights were positive:

- Outputs always positive

- Activations biased

- Gradients skewed

- Learning unstable


Centering at 0 ensures:

- Balanced signal flow

- No directional bias

- Stable gradient propagation



---

### 3️⃣ Why Not Manually Assign Signs?

We could, but:

- Random sampling is simpler

- Statistical guarantees are stronger

- Manual patterns could introduce bias


What matters is:

- Mean ≈ 0

- Variance carefully controlled


Not specific sign patterns.


---

### 4️⃣ The Real Goal of Initialization

Not just positive/negative balance.

We want:

- Var(output) ≈ Var(input)

and

- Var(gradient) ≈ constant

Across layers.

**He** and **Xavier** are mathematically derived to satisfy this.
